```python
import matplotlib.pyplot as plt
```

```python
oscar_movie = ["The Dark Night", "The Heart Locker", "The King's Speech", "The Artist", "Argo"]

year = [2008, 2009, 2010, 2011]

oscar_revenue = [1005, 170, 427, 232]
```

```python
plt.plot(year, oscar_revenue)
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")
plt.show()
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")
plt.show()
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")

plt.ylabel("Revenue in Crore")

plt.show()
```

```python
plt.plot(year, oscar_revenue, label = "Oscar Movie Revenue")

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

plt.show()
```
