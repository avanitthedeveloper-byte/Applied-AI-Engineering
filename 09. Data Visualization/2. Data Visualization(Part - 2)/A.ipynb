```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100)

plt.hist(
    scores, 
    color = "pink", 
    edgecolor = "black", 
    # bins = 15
    bins = [0, 30, 70, 90, 100]
)

plt.title("Histogram")
plt.xlabel("Scores")
plt.ylabel("# of students")
print(scores)

plt.show()
```
