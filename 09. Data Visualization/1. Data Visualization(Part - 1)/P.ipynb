```python
import matplotlib.pyplot as plt
```

```python
people = [
    "Person A",
    "Person B",
    "Person C",
    "Person D",
    "Person E",
    "Person F",
    "Person G",
    "Person H",
    "Person I",
    "Person J",
]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
```

```python
colors = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, color = colors, s = blood_pressure, marker = "^", alpha = 0.3)
plt.title("Relation between Age & Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```

```python
colors = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, cmap = "viridis", c = blood_pressure)
plt.title("Relation between Age & Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label = "Blood Pressure")
plt.savefig("Image.pdf")
plt.show()
```

```python
colors = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, cmap = "viridis", c = blood_pressure)
plt.title("Relation between Age & Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label = "Blood Pressure")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], blood_pressure[i]), xytext = (age[i]+1, blood_pressure[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_pressure), max(blood_pressure)+5)
plt.show()
```

```python
# Multiple Plots

cities = [
    "City A",
    "City B",
    "City C",
    "City D",
    "City E"
]

winter_temp = [5, 2, 10, 0, 7]
winter_humadity = [80, 75, 65, 85, 70]

summer_temp = [25, 30, 28, 35, 27]
summer_humadity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humadity, label = "Winter")
plt.scatter(summer_temp, summer_humadity, label = "Summer")

plt.title("Winter v/s Summer Data")
plt.xlabel("Temperature (℃)")
plt.ylabel("Humidity (%)")

plt.legend()

plt.show()
```
