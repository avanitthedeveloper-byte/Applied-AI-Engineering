```python
oscar_movie = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]

year = [2008, 2009, 2010, 2011, 2012]

non_oscar_revenue = [378, 2788, 829, 185, 275]
```

```python
plt.plot(year, oscar_revenue, label = "Oscar Movie")
plt.plot(year, non_oscar_revenue, label = "Non-Oscar Movie")

plt.title("Oscar Movie v/s Non Oscar Movie Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

plt.show()
```

```python
# ------------------- Format String -------------------

plt.plot(year, oscar_revenue, "o--r", label = "Oscar Movie") # Method 1
plt.plot(year, non_oscar_revenue, color = "blue", marker = ">", linestyle = ":", linewidth = 5, label = "Non-Oscar Movie") # Method 2

plt.title("Oscar Movie v/s Non Oscar Movie Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

plt.show()
```

```python
# ------------------- Styling & Saving Plots -------------------

plt.plot(year, oscar_revenue, label = "Oscar Movie")
plt.plot(year, non_oscar_revenue, label = "Non-Oscar Movie") # Method 2

plt.title("Oscar Movie v/s Non Oscar Movie Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

print(plt.style.available)

plt.style.use('seaborn-v0_8-poster')

plt.grid()

plt.show()
```

```python
# xkcd

with plt.xkcd():
    plt.plot(year, oscar_revenue, label = "Oscar Movie")
    plt.plot(year, non_oscar_revenue, label = "Non-Oscar Movie") # Method 2
    
    plt.title("Oscar Movie v/s Non Oscar Movie Revenue")
    
    plt.xlabel("Year")
    
    plt.ylabel("Revenue (in $M)")
    
    plt.legend()
    
    plt.style.use('seaborn-v0_8-poster')
    
    plt.show()
```
