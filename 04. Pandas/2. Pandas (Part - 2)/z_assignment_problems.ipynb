```python
import pandas as pd
```

```python
df = pd.read_csv("IRIS.csv")
```

```python
# 1.
    # A.
df.head(10)
df.shape
df.dtypes
df.describe()
```

```python
    # B.
df[df["petal_length"] > 4.5]
df[df["species"] == "Iris-virginica"]
```

```python
    # C.
df.groupby("species")["sepal_length"].agg(Average_Sepal_Length="mean")

df.groupby("species")["petal_width"].agg(Maximum_Petal_Length="max")

df.groupby("species")["sepal_width"].agg(Standard_Deviation_of_Sepal_Width="std")
```

```python
    # D.
df1 = df.copy()

df1["petal_ratio"] = df1["petal_length"]/df1["petal_width"]

average_petal_ratio = df1.groupby("species")["petal_ratio"].mean()

average_petal_ratio
```
