```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Corrected the range from -2*pi to +2*pi
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, -2*y, label="y = sin(x)", color="blue")

# 2. Placed the vertical dashed lines at -2*pi and 2*pi respectively
plt.axvline(0, color="black", linewidth=1.5, linestyle="--", label="x = 0")
plt.axhline(0, color="black", linewidth=1.5, linestyle="--", label="y = 0")
plt.axhline(1, color="black", linewidth=1.5, linestyle="--", label="x = 0")
plt.axhline(-1, color="black", linewidth=1.5, linestyle="--", label="y = 0")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = sin(x)")
plt.grid()
plt.legend()
plt.show()
```
