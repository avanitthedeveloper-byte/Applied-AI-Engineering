```python
import matplotlib.pyplot as plt
```

```python
oscar_movie = ["The Dark Night", "The Heart Locker", "The King's Speech", "The Artist", "Argo"]

year = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 232, 829]
```

```python
plt.plot(year, oscar_revenue)
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")
plt.show()
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")
plt.show()
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")

plt.ylabel("Revenue in Crore")

plt.show()
```

```python
plt.plot(year, oscar_revenue)

plt.title("Oscar Movie Revenue in Each Year")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend("Oscar Info")

plt.show()
```

```python
oscar_movie = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]

year = [2008, 2009, 2010, 2011, 2012]

non_oscar_revenue = [378, 2788, 829, 185, 275]
```

```python
plt.plot(year, oscar_revenue, label = "Oscar Movie")
plt.plot(year, non_oscar_revenue, label = "Non-Oscar Movie")

plt.title("Oscar Movie v/s Non Oscar Movie Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

plt.show()
```

```python
# ------------------- Format String -------------------

plt.plot(year, oscar_revenue, "o--r", label = "Oscar Movie") # Method 1
plt.plot(year, non_oscar_revenue, color = "blue", marker = ">", linestyle = ":", linewidth = 5, label = "Non-Oscar Movie") # Method 2

plt.title("Oscar Movie v/s Non Oscar Movie Revenue")

plt.xlabel("Year")

plt.ylabel("Revenue (in $M)")

plt.legend()

plt.show()
```
