```python
import matplotlib.pyplot as plt
```

```python
days = [
    "Mon",
    "Tue",
    "Wed",
    "Thu",
    "Fri"
]

cities = [
    "New Yark",
    "London",
    "Delhi",
    "Tokyo"
]

temperature = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

size = int(len(cities)/2)

fig, ax = plt.subplots(size, size, figsize = (8, 8))

temp_track = 0
for i in range(size):
    for j in range(size):
        ax[i][j].plot(days, temperature[temp_track])
        ax[i][j].set_title(cities[temp_track])
        ax[i][j].set_ylabel("<-- Temperature -->")
        temp_track = temp_track + 1

plt.suptitle("Countries Temperature Data")
fig.tight_layout()
plt.show()
```
