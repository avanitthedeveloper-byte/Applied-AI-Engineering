```python
import matplotlib.pyplot as plt
```

```python
movie = [
    "The Dark Night", 
    "The Heart Locker", 
    "The King's Speech", 
    "The Artist", 
    "Argo"]

year = [2008, 2009, 2010, 2011, 2012]

revenue = [1005, 170, 427, 232, 829]
```

```python
plt.bar(year, revenue, color = "green", label = "Revenue")

plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.title("Oscar Movie Revenue")
```

```python
plt.bar(year, revenue, color = "green")

plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.title("Oscar Movie Revenue")

for i in range(len(year)):
    plt.text(year[i], revenue[i]+10, str(revenue[i]), ha = "center")

plt.ylim(0, max(revenue)+100)
```

```python
import numpy as np

non_oscar_movie = [
    "Slumdog Millionaire", 
    "Avatar", 
    "Inception", 
    "Hugo", 
    "Lincoln"]

non_oscar_revenue = [378, 2788, 829, 185, 275]

oscar_movie = [
    "The Dark Night", 
    "The Heart Locker", 
    "The King's Speech", 
    "The Artist", 
    "Argo"]

oscar_revenue = [1005, 170, 427, 232, 829]

year = [2008, 2009, 2010, 2011, 2012]

width = 0.4
x = np.arange(len(year))

plt.xticks(x, year)
plt.bar(x-width/2, oscar_revenue, width, label = "Oscar Movie")
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Oscar Movie")

plt.title("Oscar vs Non Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```
