```python
import matplotlib.pyplot as plt

import numpy as np

import seaborn as sns
```

```python
# Q1. 
```

```python
score = np.random.normal(70, 10, 200)

plt.hist(
    score,
    edgecolor = "black"
    
)

plt.axvline(
    x = np.mean(score),
    label = "Passing Marks (#33)",
    color = "red",
    linewidth = "2",
    linestyle = "--"
)

plt.title("Histogram")
plt.xlabel("Scores")
plt.ylabel("# of students")
plt.show()
```

```python
# Q2.

sns.get_dataset_names()

pg = sns.load_dataset("penguins")

sns.scatterplot(
    data = pg,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species"
)

plt.show()
```

```python
# Q3.

tips = sns.load_dataset("tips")

sns.boxplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex"
)

plt.tight_layout()
plt.show()
```

```python
# Q4.

tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker"
)

plt.title("Tips Dataset")
plt.xlabel("Total Bill ($)") 
plt.ylabel("Tip Amount ($)")
plt.tight_layout()
plt.show()
```

```python
# Q5.

tips = sns.load_dataset("tips")

pivot_tips = tips.pivot_table(values="tip", index="day", columns="time", aggfunc="mean")

plt.figure(figsize=(8, 5))
sns.heatmap(pivot_tips, annot=True, cmap="YlGnBu", fmt=".2f", cbar_kws={"label": "Average Tip ($)"})

plt.title("Average Tip by Day and Time")
plt.xlabel("Time of Day")
plt.ylabel("Day of the Week")

plt.show()
```
