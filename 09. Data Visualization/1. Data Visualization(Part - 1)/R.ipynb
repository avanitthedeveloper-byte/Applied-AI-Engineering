```python
# 18. Customizations on Pie Charts
```

```python
import matplotlib.pyplot as plt
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
```

```python
plt.pie(amounts, labels = expenses, autopct = "%1.1f%%")

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```

```python
colors = ["darkblue", "pink", "yellow", "brown", "green"]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = colors)

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```

```python
colors = ["red", "pink", "yellow", "brown", "green"]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = colors,
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1.1,
           "linestyle": "--",
           # "alpha": 0.22,
           # "fill": False
       }
    )

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```

```python
color = ["red", "pink", "yellow", "brown", "green"]

explode = [0, 0, 0, 0.3, 0]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = color,
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1.1,
           "linestyle": "--",
           # "alpha": 0.22,
           # "fill": False
       },
        explode = explode
    )

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```

```python
color = ["red", "pink", "yellow", "brown", "green"]

explode = [0, 0, 0, 0.3, 0]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = color,
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1.1,
           "linestyle": "--",
           # "alpha": 0.22,
           # "fill": False
       },
        explode = explode,
        startangle = -180
    )

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```

```python
color = ["red", "pink", "yellow", "brown", "green"]

explode = [0, 0, 0, 0.3, 0]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = color,
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1.1,
           "linestyle": "--",
           # "alpha": 0.22,
           # "fill": False
       },
        explode = explode,
        startangle = -180,
        shadow = True
    )

plt.style.use("tableau-colorblind10")
plt.savefig("PieChart.pdf")
plt.show()
```
