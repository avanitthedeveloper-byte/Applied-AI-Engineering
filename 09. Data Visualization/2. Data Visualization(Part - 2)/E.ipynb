```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)
```

```python
plt.boxplot(
    [group1, group2],
    tick_labels = ["group1", "group2"]
)
plt.grid()
plt.show()
```

```python
plt.boxplot(
    [group1, group2],
    tick_labels = ["group1", "group2"],
    vert = False,
    showmeans = True
)
plt.grid()
plt.show()
```

```python
plt.boxplot(
    [group1, group2],
    tick_labels = ["group1", "group2"],
    vert = False,
    showmeans = True,
)
plt.grid()
plt.show()
```
