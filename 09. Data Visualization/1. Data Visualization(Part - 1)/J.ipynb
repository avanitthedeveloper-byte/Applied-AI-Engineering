```python
import matplotlib.pyplot as plt
```

```python
movie = [
    "The Dark Night", 
    "The Heart Locker", 
    "The King's Speech", 
    "The Artist", 
    "Argo"]

year = [2008, 2009, 2010, 2011, 2012]

revenue = [1005, 170, 427, 232, 829]
```

```python
plt.bar(year, revenue, color = "green", label = "Revenue")

plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.title("Oscar Movie Revenue")
```

```python
plt.bar(year, revenue, color = "green")

plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.title("Oscar Movie Revenue")

for i in range(len(year)):
    plt.text(year[i], revenue[i]+10, str(revenue[i]), ha = "center")

plt.ylim(0, max(revenue)+100)
```
