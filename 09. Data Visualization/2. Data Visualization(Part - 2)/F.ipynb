```python
import matplotlib.pyplot as plt
```

```python
# Stack Plot

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

data = [direct[0], organic[0], social[0]]
# plt.pie(data, labels = ["direct", "organic", "social"], autopct = "%1.1f%%")


plt.stackplot(
    days,
    direct,
    organic,
    social,
    labels = ["direct", "organic", "social"],
    edgecolor = "black"
)

plt.title("Stack Plot of Data")
plt.xlabel("Week's day")
plt.ylabel("# of customers")
plt.legend()

plt.show()
```
