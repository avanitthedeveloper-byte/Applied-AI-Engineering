```python
from bs4 import BeautifulSoup
```

```python
with open("scraped_data/data.html", "r") as f:
    html_content = f.read()
```

```python
soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h1")
```

```python
all_h3 = soup.find_all("h3")
```

```python
all_countires = []

for h3 in all_h3:
    name = h3.get_text(strip="true")

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip="true")

    all_countires.append([name, population])
```

```python
all_countires
```

```python
import pandas as pd

df = pd.DataFrame(all_countires, columns = ["Country", "Population"])
```

```python
df
```

```python
df.to_csv("cleaned_data/data.csv", index = False)
```
