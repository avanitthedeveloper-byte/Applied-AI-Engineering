```python
import matplotlib.pyplot as plt
```

```python
people = [
    "Person A",
    "Person B",
    "Person C",
    "Person D",
    "Person E",
    "Person F",
    "Person G",
    "Person H",
    "Person I",
    "Person J",
]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
```

```python
colors = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, color = colors, s = blood_pressure, marker = "^", alpha = 0.3)
plt.title("Relation between Age & Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```

```python
colors = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, cmap = "viridis", c = blood_pressure)
plt.title("Relation between Age & Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label = "Blood Pressure")
plt.savefig("Image.pdf")
plt.show()
```
